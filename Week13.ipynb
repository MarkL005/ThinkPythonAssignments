```python
# Mark Lucas
# Chapter 13 Journal Program

# IMPORT STATEMENTS


# FUNCTION DEFINITIONS

def write_entry():
    entry = input("Enter your journal entry: ")

    with open("journal.txt", "a") as journal_file:
        journal_file.write(entry + "\n")

    print("Entry saved.\n")


def read_journal():
    try:
        with open("journal.txt", "r") as journal_file:
            contents = journal_file.read()

            if contents == "":
                print("The journal is empty.\n")
            else:
                print("\nJournal Entries:")
                print(contents)

    except FileNotFoundError:
        print("No journal file found yet.\n")


# PROGRAM LOGIC

while True:
    choice = input("Type W to write, R to read, or Q to quit: ").upper()

    if choice == "W":
        write_entry()
    elif choice == "R":
        read_journal()
    elif choice == "Q":
        print("Goodbye.")
        break
    else:
        print("Invalid choice. Please try again.\n")
```
