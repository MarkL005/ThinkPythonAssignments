```python
import json
import os

workout_file = "boxing_workouts.json"
weight_file = "boxing_weights.json"
meal_file = "boxing_meals.json"

def load_data(file_name):
    if os.path.exists(file_name):
        with open(file_name, "r") as file:
            return json.load(file)
    return []

def save_data(file_name, data):
    with open(file_name, "w") as file:
        json.dump(data, file, indent=4)

def get_number(prompt):
    while True:
        value = input(prompt)
        try:
            return float(value)
        except ValueError:
            print("Please enter a valid number.")

workouts = load_data(workout_file)
weights = load_data(weight_file)
meals = load_data(meal_file)

def add_workout():
    date = input("Enter workout date: ").strip()
    workout_type = input("Enter workout type: ").strip()
    rounds = get_number("Enter number of rounds: ")

    if date == "" or workout_type == "":
        print("Date and workout type cannot be blank.")
        return

    workout = {
        "date": date,
        "type": workout_type,
        "rounds": rounds
    }

    workouts.append(workout)
    save_data(workout_file, workouts)
    print("Workout saved.")

def add_weight():
    date = input("Enter date: ").strip()
    weight = get_number("Enter weight: ")

    if date == "":
        print("Date cannot be blank.")
        return

    entry = {
        "date": date,
        "weight": weight
    }

    weights.append(entry)
    save_data(weight_file, weights)
    print("Weight saved.")

def add_meal():
    date = input("Enter date: ").strip()
    meal = input("Enter meal: ").strip()
    notes = input("Enter notes: ").strip()

    if date == "" or meal == "":
        print("Date and meal cannot be blank.")
        return

    entry = {
        "date": date,
        "meal": meal,
        "notes": notes
    }

    meals.append(entry)
    save_data(meal_file, meals)
    print("Meal saved.")

def view_workouts():
    if not workouts:
        print("No workouts saved.")
        return

    print("\nWorkouts:")
    for index, workout in enumerate(workouts, start=1):
        print(f"{index}. {workout['date']} | {workout['type']} | Rounds: {workout['rounds']}")

def view_weights():
    if not weights:
        print("No weights saved.")
        return

    print("\nWeights:")
    for index, weight in enumerate(weights, start=1):
        print(f"{index}. {weight['date']} | Weight: {weight['weight']}")

def view_meals():
    if not meals:
        print("No meals saved.")
        return

    print("\nMeals:")
    for index, meal in enumerate(meals, start=1):
        print(f"{index}. {meal['date']} | Meal: {meal['meal']} | Notes: {meal['notes']}")

def weekly_summary():
    total_rounds = 0

    for workout in workouts:
        total_rounds += workout["rounds"]

    print("\nWeekly Summary")
    print(f"Total workouts: {len(workouts)}")
    print(f"Total rounds: {total_rounds}")
    print(f"Weight entries: {len(weights)}")
    print(f"Meals logged: {len(meals)}")

while True:
    print("\nBoxing Training Tracker")
    print("1. Add workout")
    print("2. Add weight")
    print("3. Add meal")
    print("4. View workouts")
    print("5. View weights")
    print("6. View meals")
    print("7. Weekly summary")
    print("8. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        add_workout()
    elif choice == "2":
        add_weight()
    elif choice == "3":
        add_meal()
    elif choice == "4":
        view_workouts()
    elif choice == "5":
        view_weights()
    elif choice == "6":
        view_meals()
    elif choice == "7":
        weekly_summary()
    elif choice == "8":
        print("Goodbye.")
        break
    else:
        print("Invalid choice.")
```
