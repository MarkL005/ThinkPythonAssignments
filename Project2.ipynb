```python
import json
import os

workout_file = "boxing_workouts.json"
weight_file = "boxing_weights.json"
meal_file = "boxing_meals.json"

def load_data(file_name):
    if os.path.exists(file_name):
        with open(file_name, "r") as file:
            return json.load(file)
    return []

def save_data(file_name, data):
    with open(file_name, "w") as file:
        json.dump(data, file, indent=4)

workouts = load_data(workout_file)
weights = load_data(weight_file)
meals = load_data(meal_file)

def add_workout():
    date = input("Enter workout date: ")
    workout_type = input("Enter workout type: ")
    rounds = input("Enter number of rounds: ")

    workout = {
        "date": date,
        "type": workout_type,
        "rounds": rounds
    }

    workouts.append(workout)
    save_data(workout_file, workouts)
    print("Workout saved.")

def add_weight():
    date = input("Enter date: ")
    weight = input("Enter weight: ")

    entry = {
        "date": date,
        "weight": weight
    }

    weights.append(entry)
    save_data(weight_file, weights)
    print("Weight saved.")

def add_meal():
    date = input("Enter date: ")
    meal = input("Enter meal: ")
    notes = input("Enter notes: ")

    entry = {
        "date": date,
        "meal": meal,
        "notes": notes
    }

    meals.append(entry)
    save_data(meal_file, meals)
    print("Meal saved.")

def view_workouts():
    print("\nWorkouts:")
    for workout in workouts:
        print(workout)

def view_weights():
    print("\nWeights:")
    for weight in weights:
        print(weight)

def view_meals():
    print("\nMeals:")
    for meal in meals:
        print(meal)

while True:
    print("\nBoxing Training Tracker")
    print("1. Add workout")
    print("2. Add weight")
    print("3. Add meal")
    print("4. View workouts")
    print("5. View weights")
    print("6. View meals")
    print("7. Exit")

    choice = input("Choose an option: ")

    if choice == "1":
        add_workout()
    elif choice == "2":
        add_weight()
    elif choice == "3":
        add_meal()
    elif choice == "4":
        view_workouts()
    elif choice == "5":
        view_weights()
    elif choice == "6":
        view_meals()
    elif choice == "7":
        print("Goodbye.")
        break
    else:
        print("Invalid choice.")
```
