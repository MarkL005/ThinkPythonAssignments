```python
# Final Project: Expense Tracker
# Name: Mark Lucas

def add_expense(expenses):
    category = input("Enter category: ")
    vendor = input("Enter vendor: ")

    try:
        amount = float(input("Enter amount: "))
    except:
        print("Invalid amount")
        return

    date = input("Enter date: ")

    expense = {
        "category": category,
        "vendor": vendor,
        "amount": amount,
        "date": date
    }

    expenses.append(expense)
    print("Expense added")


def print_expense(e):
    print("Date:", e["date"])
    print("Category:", e["category"])
    print("Vendor:", e["vendor"])
    print("Amount: $", format(e["amount"], ".2f"))
    print("-----------------")


def show_all(expenses):
    total = 0

    if len(expenses) == 0:
        print("No expenses yet")
        return

    for e in expenses:
        print_expense(e)
        total += e["amount"]

    print("Total: $", format(total, ".2f"))


def by_category(expenses):
    cat = input("Enter category: ")
    total = 0
    found = False

    for e in expenses:
        if e["category"].lower() == cat.lower():
            print_expense(e)
            total += e["amount"]
            found = True

    if found:
        print("Total: $", format(total, ".2f"))
    else:
        print("None found")


def by_vendor(expenses):
    ven = input("Enter vendor: ")
    total = 0
    found = False

    for e in expenses:
        if e["vendor"].lower() == ven.lower():
            print_expense(e)
            total += e["amount"]
            found = True

    if found:
        print("Total: $", format(total, ".2f"))
    else:
        print("None found")


def reports(expenses):
    choice = ""

    while choice != "4":
        print("\nReports")
        print("1. All")
        print("2. Category")
        print("3. Vendor")
        print("4. Back")

        choice = input("Choose: ")

        if choice == "1":
            show_all(expenses)
        elif choice == "2":
            by_category(expenses)
        elif choice == "3":
            by_vendor(expenses)
        elif choice == "4":
            pass
        else:
            print("Invalid")


# MAIN PROGRAM

expenses = []
choice = ""

while choice != "3":
    print("\nMain Menu")
    print("1. Add expense")
    print("2. Reports")
    print("3. Quit")

    choice = input("Choose: ")

    if choice == "1":
        add_expense(expenses)
    elif choice == "2":
        reports(expenses)
    elif choice == "3":
        print("Goodbye")
    else:
        print("Invalid choice")
```
