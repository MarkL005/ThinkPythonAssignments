```python
# STRUCTURE & PSEUDOCODE

# IMPORT STATEMENTS
# No imports needed for this program

# CLASS DEFINITIONS
# Create a Book class
# The Book class should keep track of:
# - title
# - author
# - whether the book is available

# The Book class should have methods to:
# - display the book as a string
# - check out the book
# - return the book

# PROGRAM LOGIC
# Create at least two Book objects
# Print both books
# Check out one book
# Print both books again to show the change


class Book:
    def __init__(self, title, author):
        self.title = title
        self.author = author
        self.available = True

    def __str__(self):
        if self.available:
            status = "Available"
        else:
            status = "Checked Out"
        return f"Title: {self.title}, Author: {self.author}, Status: {status}"

    def checkout(self):
        self.available = False

    def return_book(self):
        self.available = True


# PROGRAM LOGIC
book1 = Book("The Great Gatsby", "F. Scott Fitzgerald")
book2 = Book("To Kill a Mockingbird", "Harper Lee")

print("Before checkout:")
print(book1)
print(book2)

book1.checkout()

print()
print("After checkout:")
print(book1)
print(book2)
```
